## Here we perform experiments to understand exponential moving average and Momentum Optimizer 

```python
import numpy as np 
import matplotlib.pyplot as plt 
import math
import random 
```

```python
np.random.seed(42)
```

```python
x = np.linspace(start=0, stop=10, num=50)
y = np.cos(x) + 0.3 * np.random.randn(len(x))
```

```python
plt.scatter(x,y,c="r", label="cosine")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
'''
Our goal is to approximate the above function using an exponential weighted moving average with and without bias correction 
'''
```

```python
# 1. Exponential weighted moving average without bias correction 
```

```python
data = y
state = [0]
beta = 0.9 

assert 0.0 <= beta <= 1 

for i in range(1, len(data)):
    state.append(
        beta * state[i - 1] + (1 - beta) * data[i]
    )
```

```python
# 2. Exponential weighted moving average with Bias Correction 
```

```python
state_2 = [0]
beta = 0.7 

assert 0.0 <= beta <= 1 

for i in range(1, len(data)):
    v = beta * state[i-1] + (1-beta) * data[i]
    v_corrected = v / (1 - beta**i)

    state_2.append(v_corrected)
```

```python
plt.plot(x,y,c="r", label="cosine")
plt.plot(x,state, c="b", label = "ExpMA")
plt.plot(x,state_2, c = "g", label="bias correction")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
